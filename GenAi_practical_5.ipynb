```python
import numpy as np

print("--- Simple Attention Mechanism Explanation ---")

# 1. Encoder Outputs
# Each word has a 2-dimensional feature vector
encoder_outputs = np.array([
    [0.1, 0.2],   # 'I'
    [0.5, 0.3],   # 'am'
    [0.2, 0.8],   # 'a'
    [0.9, 0.1]    # 'student'
])

print(f"\nEncoder Outputs (4 words, 2 features each):\n{encoder_outputs}")


# 2. Dummy Decoder Hidden State
# What the decoder is currently trying to generate
# Let's say the decoder is trying to generate a word related to 'learning'
decoder_hidden_state = np.array([0.7, 0.4])

print(
    f"\nDecoder Hidden State (what we are looking for): "
    f"{decoder_hidden_state}"
)


# 3. Calculate Alignment Scores or Energy
# How well each encoder output matches the decoder state
# A simple dot product can show similarity

alignment_scores = np.dot(
    encoder_outputs,
    decoder_hidden_state
)

print(
    f"\nAlignment Scores (similarity of each encoder output "
    f"to decoder state): {alignment_scores}"
)


# 4. Apply Softmax to get Attention Weights
# Probabilities summing to 1

def softmax(x):
    e_x = np.exp(x - np.max(x))  # Numerical stability
    return e_x / e_x.sum(axis=0)


attention_weights = softmax(alignment_scores)

print(
    f"\nAttention Weights (how much to 'pay attention' "
    f"to each word): {attention_weights}"
)

print(
    "(Notice that the word 'student' (index 3) got a higher "
    "weight, meaning it's more relevant to 'learning')"
)


# 5. Compute the Context Vector
# This is a weighted sum of encoder outputs,
# focusing on the most relevant parts

context_vector = np.sum(
    encoder_outputs * attention_weights[:, np.newaxis],
    axis=0
)

print(
    f"\nContext Vector (weighted sum of encoder outputs, "
    f"paying attention): {context_vector}"
)


print("\n--- How it works ---")
print("1. Encoder creates representations for each word.")
print("2. Decoder has a hidden state representing what it needs.")
print("3. Alignment scores measure similarity.")
print("4. Softmax converts scores into attention weights.")
print("5. Weighted sum creates the context vector.")
```
